```python
#accessing last year's (2025) data from the betting database
%sql SELECT * FROM business.crishbet_wa_1550;
```

```python
#checking the specific activities recorded in the year
%sql SELECT DISTINCT event_name FROM business.crishbet_wa_1550
```

```python
#checking which activity had the most occurrence
%sql SELECT DISTINCT event_name, COUNT(event_name) AS recorded FROM business.crishbet_wa_1550 GROUP BY event_name ORDER BY recorded DESC;
```

```python
#checking all records where bet was staked.
%sql SELECT * FROM business.crishbet_wa_1550 WHERE bet_type <> '';
```

```python
#checking all records where deposits were made.
%sql SELECT * FROM business.crishbet_wa_1550 WHERE event_name = 'Deposit Made';
```

```python
#checking all records where bet staked is null.
%sql SELECT * FROM business.crishbet_wa_1550 WHERE stake_amount = '';
```

```python
#identifying which platforms customers used the most to interact with the company's product.
%sql SELECT device, COUNT(device) AS recorded FROM business.crishbet_wa_1550 GROUP BY device ORDER BY recorded DESC; 
```

```python
#identifying which activities recorded with no devices to follow-up.
%sql SELECT event_name, COUNT(event_name) AS device_unknown FROM business.crishbet_wa_1550 WHERE device IS NULL OR device = '' GROUP BY event_name;
```

```python
#what type of bets were staked in 2025
%sql SELECT DISTINCT bet_type FROM business.crishbet_wa_1550;
```

```python
#identifying most preferred sports with bet types identified.
%sql SELECT sport, COUNT(sport) AS counted FROM business.crishbet_wa_1550 WHERE bet_type <> '' GROUP BY sport ORDER BY COUNT(sport) DESC;
```

```python
#identifying top 10 bet types and their sports where customers preferred to stake high amounts for the year.
%sql SELECT bet_type, sport, stake_amount AS counted FROM business.crishbet_wa_1550 WHERE bet_type <> '' ORDER BY stake_amount DESC LIMIT 10;
```

```python
#identifying countries where customers who staked the bet were located.
%sql SELECT country, COUNT(country) AS counted FROM business.crishbet_wa_1550 WHERE bet_type <> '' GROUP BY country ORDER BY COUNT(country) DESC;
```

```python
#checking all records where user ids appeared more than once.
%sql SELECT user_id, COUNT(user_id) AS appeared FROM business.crishbet_wa_1550 GROUP BY user_id HAVING COUNT(user_id) > 2 ORDER BY appeared DESC;
```
